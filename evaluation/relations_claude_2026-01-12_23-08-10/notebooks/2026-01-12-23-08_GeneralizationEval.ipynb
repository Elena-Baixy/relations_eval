## Repository Overview

This repository implements **Linear Relational Embeddings (LRE)** from the paper "Linearity of Relation Decoding in Transformer Language Models" (Hernandez et al., 2023).

### Key Findings:
1. For many relations, the highly non-linear decoding procedure in transformers can be approximated by a simple **linear transformation** (LRE) on the subject representation at intermediate layers
2. LRE(s) = Ws + b maps subject representation s to object representation o
3. The transformation can be computed from the LM Jacobian

### Models Used in Original Work:
- GPT-J (6B)
- GPT-2-XL 
- LLaMA-13B

### Dataset:
- 47 relations across factual, commonsense, linguistic, and bias categories
- Over 10k subject-object pairs

# GT1: Model Generalization Test

Testing whether the Linear Relational Embedding (LRE) finding generalizes to a NEW model not used in the original work.

**Original models**: GPT-J-6B, GPT-2-XL, LLaMA-13B

**New model to test**: Pythia-2.8B (not used in original paper, similar architecture to GPT-J)

The key finding is that relational knowledge can be decoded via linear transformations on subject representations. We will test if this holds on Pythia.

```python
# GT1: Test model generalization using Pythia-2.8B (not used in original work)
# We'll implement the LRE approach on this new model

from transformers import AutoModelForCausalLM, AutoTokenizer
import torch

# Load Pythia-2.8B (not used in original paper)
model_name = "EleutherAI/pythia-2.8b"
print(f"Loading {model_name}...")

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Using device: {device}")

tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForCausalLM.from_pretrained(model_name, torch_dtype=torch.float16).to(device)
model.eval()
print(f"Model loaded successfully on {device}")
```

```python
# Check if model is loaded
print(f"Model: {type(model).__name__}")
print(f"Number of layers: {model.config.num_hidden_layers}")
print(f"Hidden size: {model.config.hidden_size}")
```

```python
# Wait for model loading and check status
import time
time.sleep(2)
print(f"Model type: {type(model).__name__}")
print(f"Model is on: {next(model.parameters()).device}")
```

```python
# Test basic model inference
test_prompt = "The capital city of France is"
inputs = tokenizer(test_prompt, return_tensors="pt").to(device)
with torch.no_grad():
    outputs = model.generate(**inputs, max_new_tokens=5, do_sample=False)
result = tokenizer.decode(outputs[0], skip_special_tokens=True)
print(f"Test prompt: {test_prompt}")
print(f"Model output: {result}")
```
