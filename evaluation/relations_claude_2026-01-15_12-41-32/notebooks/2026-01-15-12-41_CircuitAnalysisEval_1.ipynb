```python
# Block 6: Analyze next-token predictions (sarcastic)
print("=" * 60)
print("BLOCK 6: Analyze next-token predictions (sarcastic sentences)")
print("=" * 60)

try:
    def get_top_predictions(sentence, k=10):
        """Get top k next token predictions for a sentence."""
        tokens = model.to_tokens(sentence)
        with torch.no_grad():
            logits = model(tokens)
        # Get logits for last position
        last_logits = logits[0, -1, :]
        probs = F.softmax(last_logits, dim=-1)
        top_probs, top_indices = torch.topk(probs, k)
        top_tokens = [model.tokenizer.decode(idx.item()) for idx in top_indices]
        return list(zip(top_tokens, top_probs.cpu().numpy()))

    print("Next-token predictions for sarcastic vs literal sentences:")
    print("=" * 70)

    for i, (prompt, expected_neg, expected_pos) in enumerate(sarcasm_with_continuation[:5]):
        print(f"\n[SARCASTIC] {prompt}")
        predictions = get_top_predictions(prompt)
        print("  Top predictions:", predictions[:5])
        
        # Check if negative or positive continuation is more likely
        neg_prob = 0
        pos_prob = 0
        for token, prob in predictions[:20]:
            if any(neg in token.lower() for neg in ['awful', 'bad', 'terrible', 'frustrat', 'annoy', 'upset', 'sad', 'hate']):
                neg_prob += prob
            if any(pos in token.lower() for pos in ['great', 'amaz', 'happy', 'excit', 'good', 'proud', 'love', 'wonderful']):
                pos_prob += prob
        print(f"  Negative sentiment prob: {neg_prob:.4f}, Positive sentiment prob: {pos_prob:.4f}")
    
    # Evaluation
    code_cells_info[6]['description'] = "Analyze next-token predictions for sarcastic sentences"
    code_cells_info[6]['runnable'] = 'Y'
    code_cells_info[6]['correct_implementation'] = 'Y'
    code_cells_info[6]['redundant'] = 'N'
    code_cells_info[6]['irrelevant'] = 'N'
    print("\n✓ Block 6 executed successfully")
except Exception as e:
    code_cells_info[6]['runnable'] = 'N'
    code_cells_info[6]['error_note'] = str(e)
    print(f"\n✗ Block 6 failed: {e}")
```

```python
# Block 8: Create cleaner experimental setup
print("=" * 60)
print("BLOCK 8: Create cleaner experimental setup")
print("=" * 60)

try:
    # Create prompts where we can measure the difference in sentiment prediction
    sarcasm_prompts = [
        "Oh great, another meeting. I feel",  # sarcastic, expects negative
        "Oh great, I got a raise. I feel",    # literal, expects positive
        "Wonderful, more work to do. I feel", # sarcastic, expects negative
        "Wonderful, I won the lottery. I feel", # literal, expects positive
        "Just what I needed, a flat tire. I feel", # sarcastic, expects negative
        "Just what I needed, a vacation. I feel",  # literal, expects positive
        "Perfect, I lost my wallet. I feel",  # sarcastic, expects negative
        "Perfect, I found true love. I feel", # literal, expects positive
        "Fantastic, the car broke down. I feel", # sarcastic, expects negative
        "Fantastic, I got the job. I feel",   # literal, expects positive
    ]

    # Define positive and negative sentiment words to probe
    positive_words = [" happy", " good", " great", " excited", " wonderful", " amazing", " pleased", " glad", " thrilled", " joyful"]
    negative_words = [" bad", " sad", " terrible", " awful", " frustrated", " annoyed", " upset", " angry", " miserable", " disappointed"]

    def get_sentiment_probs(prompt, positive_words, negative_words):
        """Get probability mass on positive vs negative sentiment words."""
        tokens = model.to_tokens(prompt)
        with torch.no_grad():
            logits = model(tokens)
        last_logits = logits[0, -1, :]
        probs = F.softmax(last_logits, dim=-1)
        
        pos_prob = 0
        neg_prob = 0
        
        for word in positive_words:
            word_token = model.to_tokens(word, prepend_bos=False)[0, 0]
            pos_prob += probs[word_token].item()
        
        for word in negative_words:
            word_token = model.to_tokens(word, prepend_bos=False)[0, 0]
            neg_prob += probs[word_token].item()
        
        return pos_prob, neg_prob

    print("Sentiment analysis for sarcasm prompts:")
    print("=" * 70)
    for i, prompt in enumerate(sarcasm_prompts):
        pos_prob, neg_prob = get_sentiment_probs(prompt, positive_words, negative_words)
        sentiment_type = "SARCASTIC" if i % 2 == 0 else "LITERAL"
        print(f"[{sentiment_type}] {prompt}")
        print(f"  Positive: {pos_prob:.4f}, Negative: {neg_prob:.4f}")
    
    # Evaluation
    code_cells_info[8]['description'] = "Create cleaner experimental setup with sentiment probes"
    code_cells_info[8]['runnable'] = 'Y'
    code_cells_info[8]['correct_implementation'] = 'Y'
    code_cells_info[8]['redundant'] = 'N'
    code_cells_info[8]['irrelevant'] = 'N'
    print("\n✓ Block 8 executed successfully")
except Exception as e:
    code_cells_info[8]['runnable'] = 'N'
    code_cells_info[8]['error_note'] = str(e)
    print(f"\n✗ Block 8 failed: {e}")
```

```python
# Block 9: Phase 3 - Collect activations for minimal pairs
print("=" * 60)
print("BLOCK 9: Collect activations for minimal pairs")
print("=" * 60)

try:
    # Create cleaner minimal pairs with same structure
    minimal_pairs = [
        ("Oh great, another boring meeting", "Oh great, a wonderful surprise"),
        ("Wonderful, more tedious paperwork", "Wonderful, a fantastic opportunity"),
        ("Perfect, I lost my wallet today", "Perfect, I found twenty dollars"),
        ("Fantastic, the car broke down", "Fantastic, the team won today"),
        ("Amazing, I have to work late", "Amazing, I got a promotion"),
    ]

    def collect_activations(prompt):
        """Collect all residual stream activations for a prompt."""
        tokens = model.to_tokens(prompt)
        with torch.no_grad():
            _, cache = model.run_with_cache(tokens)
        return cache

    # Collect activations for all pairs
    print("Collecting activations for minimal pairs...")
    activations_sarcastic = []
    activations_literal = []

    for sarc, lit in minimal_pairs:
        cache_sarc = collect_activations(sarc)
        cache_lit = collect_activations(lit)
        activations_sarcastic.append(cache_sarc)
        activations_literal.append(cache_lit)
        
    print(f"Collected activations for {len(minimal_pairs)} pairs")
    print(f"Cache keys available: {len(cache_sarc.keys())} keys")
    
    # Evaluation
    code_cells_info[9]['description'] = "Collect activations for minimal pairs"
    code_cells_info[9]['runnable'] = 'Y'
    code_cells_info[9]['correct_implementation'] = 'Y'
    code_cells_info[9]['redundant'] = 'N'
    code_cells_info[9]['irrelevant'] = 'N'
    print("\n✓ Block 9 executed successfully")
except Exception as e:
    code_cells_info[9]['runnable'] = 'N'
    code_cells_info[9]['error_note'] = str(e)
    print(f"\n✗ Block 9 failed: {e}")
```

```python
# Block 10: Analyze residual stream divergence across layers
print("=" * 60)
print("BLOCK 10: Analyze residual stream divergence across layers")
print("=" * 60)

try:
    def compute_cosine_similarity(vec1, vec2):
        """Compute cosine similarity between two vectors."""
        return F.cosine_similarity(vec1.flatten().unsqueeze(0), vec2.flatten().unsqueeze(0)).item()

    def compute_l2_distance(vec1, vec2):
        """Compute L2 distance between two vectors."""
        return torch.norm(vec1 - vec2).item()

    # Analyze residual stream at each layer
    print("Residual stream divergence (sarcastic vs literal) at last token:")
    print("=" * 70)
    print(f"{'Layer':<10} {'Cosine Sim':<15} {'L2 Distance':<15}")
    print("-" * 70)

    layer_similarities = []
    layer_distances = []

    for layer in range(model.cfg.n_layers + 1):  # +1 for final residual
        if layer == 0:
            key = "blocks.0.hook_resid_pre"
        elif layer == model.cfg.n_layers:
            key = f"blocks.{layer-1}.hook_resid_post"
        else:
            key = f"blocks.{layer}.hook_resid_pre"
        
        cosine_sims = []
        l2_dists = []
        
        for i, (sarc, lit) in enumerate(minimal_pairs):
            # Get last token activations
            act_sarc = activations_sarcastic[i][key][0, -1, :]
            act_lit = activations_literal[i][key][0, -1, :]
            
            cosine_sims.append(compute_cosine_similarity(act_sarc, act_lit))
            l2_dists.append(compute_l2_distance(act_sarc, act_lit))
        
        avg_cosine = np.mean(cosine_sims)
        avg_l2 = np.mean(l2_dists)
        layer_similarities.append(avg_cosine)
        layer_distances.append(avg_l2)
        
        print(f"Layer {layer:<4} {avg_cosine:.4f}          {avg_l2:.4f}")

    print("\nLargest divergence at layers:", np.argsort(layer_similarities)[:3])
    
    # Evaluation
    code_cells_info[10]['description'] = "Analyze residual stream divergence across layers"
    code_cells_info[10]['runnable'] = 'Y'
    code_cells_info[10]['correct_implementation'] = 'Y'
    code_cells_info[10]['redundant'] = 'N'
    code_cells_info[10]['irrelevant'] = 'N'
    print("\n✓ Block 10 executed successfully")
except Exception as e:
    code_cells_info[10]['runnable'] = 'N'
    code_cells_info[10]['error_note'] = str(e)
    print(f"\n✗ Block 10 failed: {e}")
```

```python
# Block 12: Define sentiment logit diff function and get baseline
print("=" * 60)
print("BLOCK 12: Define sentiment logit diff function and baseline")
print("=" * 60)

try:
    def get_sentiment_logit_diff(prompt, positive_tokens, negative_tokens):
        """Get the difference between positive and negative sentiment logits."""
        tokens = model.to_tokens(prompt)
        with torch.no_grad():
            logits = model(tokens)
        last_logits = logits[0, -1, :]
        
        pos_logit = torch.mean(torch.stack([last_logits[t] for t in positive_tokens]))
        neg_logit = torch.mean(torch.stack([last_logits[t] for t in negative_tokens]))
        
        return (pos_logit - neg_logit).item()

    # Get token IDs for sentiment words
    positive_tokens = [model.to_tokens(w, prepend_bos=False)[0, 0].item() for w in positive_words]
    negative_tokens = [model.to_tokens(w, prepend_bos=False)[0, 0].item() for w in negative_words]

    print("Positive token IDs:", positive_tokens[:5])
    print("Negative token IDs:", negative_tokens[:5])

    # Test baseline
    test_sarcastic = "Perfect, I lost my wallet. I feel"
    test_literal = "Perfect, I found some money. I feel"

    baseline_sarc = get_sentiment_logit_diff(test_sarcastic, positive_tokens, negative_tokens)
    baseline_lit = get_sentiment_logit_diff(test_literal, positive_tokens, negative_tokens)

    print(f"\nBaseline sentiment logit diff (pos - neg):")
    print(f"  Sarcastic: {baseline_sarc:.4f}")
    print(f"  Literal: {baseline_lit:.4f}")
    print(f"  Gap: {baseline_lit - baseline_sarc:.4f}")
    
    # Evaluation
    code_cells_info[12]['description'] = "Define sentiment logit diff function and get baseline"
    code_cells_info[12]['runnable'] = 'Y'
    code_cells_info[12]['correct_implementation'] = 'Y'
    code_cells_info[12]['redundant'] = 'N'
    code_cells_info[12]['irrelevant'] = 'N'
    print("\n✓ Block 12 executed successfully")
except Exception as e:
    code_cells_info[12]['runnable'] = 'N'
    code_cells_info[12]['error_note'] = str(e)
    print(f"\n✗ Block 12 failed: {e}")
```

```python
# Block 13: Define head patching function
print("=" * 60)
print("BLOCK 13: Define head patching function")
print("=" * 60)

try:
    def patch_head_and_measure(sarcastic_prompt, literal_prompt, layer, head, positive_tokens, negative_tokens):
        """Patch attention head output from literal into sarcastic and measure change."""
        
        # Get clean activations from literal
        tokens_lit = model.to_tokens(literal_prompt)
        _, cache_lit = model.run_with_cache(tokens_lit)
        
        # Define patching hook
        def patch_hook(activation, hook):
            # activation shape: [batch, seq, n_heads, d_head]
            # Patch only the specific head at last position
            activation[0, -1, head, :] = cache_lit[hook.name][0, -1, head, :]
            return activation
        
        # Run with patching
        tokens_sarc = model.to_tokens(sarcastic_prompt)
        hook_name = f"blocks.{layer}.attn.hook_z"
        
        with torch.no_grad():
            patched_logits = model.run_with_hooks(
                tokens_sarc,
                fwd_hooks=[(hook_name, patch_hook)]
            )
        
        last_logits = patched_logits[0, -1, :]
        pos_logit = torch.mean(torch.stack([last_logits[t] for t in positive_tokens]))
        neg_logit = torch.mean(torch.stack([last_logits[t] for t in negative_tokens]))
        
        return (pos_logit - neg_logit).item()

    # Test patching one head
    test_result = patch_head_and_measure(
        test_sarcastic, test_literal, 
        layer=5, head=0, 
        positive_tokens=positive_tokens, 
        negative_tokens=negative_tokens
    )
    print(f"After patching L5H0: {test_result:.4f} (baseline: {baseline_sarc:.4f})")
    print(f"Change: {test_result - baseline_sarc:.4f}")
    
    # Evaluation
    code_cells_info[13]['description'] = "Define head patching function"
    code_cells_info[13]['runnable'] = 'Y'
    code_cells_info[13]['correct_implementation'] = 'Y'
    code_cells_info[13]['redundant'] = 'N'
    code_cells_info[13]['irrelevant'] = 'N'
    print("\n✓ Block 13 executed successfully")
except Exception as e:
    code_cells_info[13]['runnable'] = 'N'
    code_cells_info[13]['error_note'] = str(e)
    print(f"\n✗ Block 13 failed: {e}")
```

```python
# Block 14: Patch all attention heads and collect results
print("=" * 60)
print("BLOCK 14: Patch all attention heads and collect results")
print("=" * 60)

try:
    print("Running activation patching for all attention heads...")
    print("This will take a moment...")

    head_patch_effects = np.zeros((model.cfg.n_layers, model.cfg.n_heads))

    for layer in tqdm(range(model.cfg.n_layers)):
        for head in range(model.cfg.n_heads):
            patched_result = patch_head_and_measure(
                test_sarcastic, test_literal,
                layer, head,
                positive_tokens, negative_tokens
            )
            # Effect = how much patching moves toward literal behavior
            head_patch_effects[layer, head] = patched_result - baseline_sarc

    print("\nTop 10 attention heads by patching effect:")
    flat_indices = np.argsort(head_patch_effects.flatten())[::-1]
    for idx in flat_indices[:10]:
        layer = idx // model.cfg.n_heads
        head = idx % model.cfg.n_heads
        effect = head_patch_effects[layer, head]
        print(f"  a{layer}.h{head}: {effect:.4f}")
    
    # Evaluation
    code_cells_info[14]['description'] = "Patch all attention heads and collect results"
    code_cells_info[14]['runnable'] = 'Y'
    code_cells_info[14]['correct_implementation'] = 'Y'
    code_cells_info[14]['redundant'] = 'N'
    code_cells_info[14]['irrelevant'] = 'N'
    print("\n✓ Block 14 executed successfully")
except Exception as e:
    code_cells_info[14]['runnable'] = 'N'
    code_cells_info[14]['error_note'] = str(e)
    print(f"\n✗ Block 14 failed: {e}")
```

```python
# Block 16: Patch MLPs
print("=" * 60)
print("BLOCK 16: Patch MLPs")
print("=" * 60)

try:
    def patch_mlp_and_measure(sarcastic_prompt, literal_prompt, layer, positive_tokens, negative_tokens):
        """Patch MLP output from literal into sarcastic and measure change."""
        
        # Get clean activations from literal
        tokens_lit = model.to_tokens(literal_prompt)
        _, cache_lit = model.run_with_cache(tokens_lit)
        
        # Define patching hook
        def patch_hook(activation, hook):
            # activation shape: [batch, seq, d_model]
            # Patch at last position
            activation[0, -1, :] = cache_lit[hook.name][0, -1, :]
            return activation
        
        # Run with patching
        tokens_sarc = model.to_tokens(sarcastic_prompt)
        hook_name = f"blocks.{layer}.hook_mlp_out"
        
        with torch.no_grad():
            patched_logits = model.run_with_hooks(
                tokens_sarc,
                fwd_hooks=[(hook_name, patch_hook)]
            )
        
        last_logits = patched_logits[0, -1, :]
        pos_logit = torch.mean(torch.stack([last_logits[t] for t in positive_tokens]))
        neg_logit = torch.mean(torch.stack([last_logits[t] for t in negative_tokens]))
        
        return (pos_logit - neg_logit).item()

    print("Running activation patching for all MLPs...")
    mlp_patch_effects = []

    for layer in tqdm(range(model.cfg.n_layers)):
        patched_result = patch_mlp_and_measure(
            test_sarcastic, test_literal,
            layer,
            positive_tokens, negative_tokens
        )
        effect = patched_result - baseline_sarc
        mlp_patch_effects.append(effect)

    print("\nMLP patching effects by layer:")
    for layer, effect in enumerate(mlp_patch_effects):
        bar = "+" * int(abs(effect) * 10) if effect > 0 else "-" * int(abs(effect) * 10)
        print(f"  m{layer}: {effect:+.4f} {bar}")
    
    # Evaluation
    code_cells_info[16]['description'] = "Patch MLPs"
    code_cells_info[16]['runnable'] = 'Y'
    code_cells_info[16]['correct_implementation'] = 'Y'
    code_cells_info[16]['redundant'] = 'N'
    code_cells_info[16]['irrelevant'] = 'N'
    print("\n✓ Block 16 executed successfully")
except Exception as e:
    code_cells_info[16]['runnable'] = 'N'
    code_cells_info[16]['error_note'] = str(e)
    print(f"\n✗ Block 16 failed: {e}")
```

```python
# Block 17: Validate across multiple sentence pairs
print("=" * 60)
print("BLOCK 17: Validate across multiple sentence pairs")
print("=" * 60)

try:
    print("Validating across multiple sentence pairs...")

    all_head_effects = []
    all_mlp_effects = []

    validation_pairs = [
        ("Oh great, another boring meeting. I feel", "Oh great, a wonderful surprise. I feel"),
        ("Wonderful, more tedious paperwork. I feel", "Wonderful, a fantastic opportunity. I feel"),
        ("Perfect, I lost my wallet today. I feel", "Perfect, I found twenty dollars. I feel"),
        ("Fantastic, the car broke down. I feel", "Fantastic, the team won today. I feel"),
        ("Amazing, I have to work overtime. I feel", "Amazing, I got a raise today. I feel"),
    ]

    for sarc, lit in tqdm(validation_pairs):
        # Get baselines
        baseline_s = get_sentiment_logit_diff(sarc, positive_tokens, negative_tokens)
        
        # Patch heads
        pair_head_effects = np.zeros((model.cfg.n_layers, model.cfg.n_heads))
        for layer in range(model.cfg.n_layers):
            for head in range(model.cfg.n_heads):
                result = patch_head_and_measure(sarc, lit, layer, head, positive_tokens, negative_tokens)
                pair_head_effects[layer, head] = result - baseline_s
        all_head_effects.append(pair_head_effects)
        
        # Patch MLPs
        pair_mlp_effects = []
        for layer in range(model.cfg.n_layers):
            result = patch_mlp_and_measure(sarc, lit, layer, positive_tokens, negative_tokens)
            pair_mlp_effects.append(result - baseline_s)
        all_mlp_effects.append(pair_mlp_effects)

    # Average effects
    avg_head_effects = np.mean(all_head_effects, axis=0)
    avg_mlp_effects = np.mean(all_mlp_effects, axis=0)

    print("\nTop attention heads (averaged across pairs):")
    flat_indices = np.argsort(avg_head_effects.flatten())[::-1]
    for idx in flat_indices[:10]:
        layer = idx // model.cfg.n_heads
        head = idx % model.cfg.n_heads
        effect = avg_head_effects[layer, head]
        print(f"  a{layer}.h{head}: {effect:.4f}")
    
    # Evaluation
    code_cells_info[17]['description'] = "Validate across multiple sentence pairs"
    code_cells_info[17]['runnable'] = 'Y'
    code_cells_info[17]['correct_implementation'] = 'Y'
    code_cells_info[17]['redundant'] = 'N'
    code_cells_info[17]['irrelevant'] = 'N'
    print("\n✓ Block 17 executed successfully")
except Exception as e:
    code_cells_info[17]['runnable'] = 'N'
    code_cells_info[17]['error_note'] = str(e)
    print(f"\n✗ Block 17 failed: {e}")
```
