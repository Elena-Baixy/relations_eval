# Code Evaluation: Linearity of Relation Decoding in Transformer LMs

## Evaluation Mode: No-Execution Code Evaluation

This evaluation is based **solely on static inspection** of the code files. No code execution is performed.

### Source of Project Goal

The evaluation uses:
1. **Plan file** (`plan.md`): Defines the objective to investigate how transformer LMs represent and decode relational knowledge using Linear Relational Embeddings (LRE).
2. **CodeWalkthrough file** (`CodeWalkthrough.md`): Identifies the main demo notebooks as the core analysis entry points.

### Project Goal Summary

The project aims to:
1. Extract Linear Relational Embeddings (LREs) by computing the mean Jacobian W and bias b
2. Evaluate LRE faithfulness (whether LRE predictions match the full transformer)
3. Evaluate LRE causality (whether LRE edits change model predictions)
4. Test on GPT-J, GPT-2-XL, and LLaMA-13B using 47 relations with 10k+ subject-object pairs

## 1. Block-Level Evaluation Table

The following table evaluates each code block/function in the core implementation files. 

### Evaluation Criteria:
- **Runnable (Y/N)**: Syntactically valid, imports available, no obvious runtime blockers
- **Correct-Implementation (Y/N)**: Logic correctly implements the described computation
- **Redundant (Y/N)**: Duplicates another block's computation
- **Irrelevant (Y/N)**: Does not contribute to the project goal
